```python
import pandas as pd
from pathlib import Path

# Folder containing the Olist CSV files
DATA_DIR = Path("/Users/akshatsinha/Olist/data")

# Find all CSV files
csv_files = sorted(DATA_DIR.glob("*.csv"))

print(f"Found {len(csv_files)} CSV files\n")
print("=" * 100)

for file in csv_files:
    print(f"\n📄 {file.name}")
    print("-" * 100)

    # Load CSV
    df = pd.read_csv(file)

    # Basic information
    print(f"Rows    : {len(df):,}")
    print(f"Columns : {len(df.columns)}")

    # Column information
    print("\nColumns & Data Types:")
    print(df.dtypes.to_string())

    # Missing values
    print("\nMissing Values:")
    missing = df.isnull().sum()
    missing_pct = (missing / len(df) * 100).round(2)

    missing_table = pd.DataFrame({
        "Missing": missing,
        "Missing %": missing_pct
    })

    print(missing_table[missing_table["Missing"] > 0].to_string())

    # Sample rows
    print("\nSample Rows:")
    print(df.head(3).to_string(index=False))

    print("\n" + "=" * 100)
```

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("/Users/akshatsinha/Olist/data")
OUTPUT_DIR = Path("olist_inspection")
OUTPUT_DIR.mkdir(exist_ok=True)

csv_files = sorted(DATA_DIR.glob("*.csv"))

summary = []

for file in csv_files:
    df = pd.read_csv(file)

    # Missing values
    missing = df.isnull().sum()
    missing_pct = (missing / len(df) * 100).round(2)

    # Save detailed report
    report_file = OUTPUT_DIR / f"{file.stem}_report.txt"

    with open(report_file, "w", encoding="utf-8") as f:

        f.write(f"FILE: {file.name}\n")
        f.write("=" * 100 + "\n\n")

        f.write(f"ROWS: {len(df):,}\n")
        f.write(f"COLUMNS: {len(df.columns)}\n\n")

        f.write("COLUMNS & DATA TYPES\n")
        f.write("-" * 100 + "\n")
        f.write(df.dtypes.to_string())
        f.write("\n\n")

        f.write("MISSING VALUES\n")
        f.write("-" * 100 + "\n")

        missing_table = pd.DataFrame({
            "Missing": missing,
            "Missing %": missing_pct
        })

        f.write(missing_table.to_string())
        f.write("\n\n")

        f.write("UNIQUE VALUES\n")
        f.write("-" * 100 + "\n")

        for col in df.columns:
            f.write(f"{col}: {df[col].nunique():,}\n")

        f.write("\nSAMPLE ROWS\n")
        f.write("-" * 100 + "\n")
        f.write(df.head(5).to_string(index=False))

    # Summary
    summary.append({
        "file": file.name,
        "rows": len(df),
        "columns": len(df.columns)
    })

# Save overall summary
summary_df = pd.DataFrame(summary)
summary_df.to_csv(OUTPUT_DIR / "dataset_summary.csv", index=False)

print("Inspection complete!")
print(f"Created folder: {OUTPUT_DIR}")
print("\nFiles created:")

for f in sorted(OUTPUT_DIR.iterdir()):
    print(f"  - {f.name}")
```
